# 0.8 - Data Columns

Saves the unique values of each feature to `cols_values.json`.

```python
import json

import pandas as pd

ROOT = '../../'

X_train = pd.read_csv(f'{ROOT}data/final/dataset-v0.2-x-train.csv', index_col=0)
cols_values = {column: X_train[column].dropna().unique().tolist() for column in X_train.columns}

with open('./cols_values.json', 'w', encoding='utf-8') as file:
    json.dump(cols_values, file, indent=2, ensure_ascii=False)
```
