# 0.3 - Merge RegBL and Radon

This notebooks' objective is to merge two datasets in their `processed` versions : 

- `01_OFSP_radon_bdd`
- `03_OFS_regbl`

## Constants

```python
ROOT = '../../'
```

## Load the data

```python
df_radon = pd.read_csv(f'{ROOT}data/processed/01_OFSP_radon_bdd.csv', index_col=0)
df_regbl = pd.read_csv(f'{ROOT}data/processed/03_OFS_regbl.csv', index_col=0)
```

## Merge the data

```python
df_radon.head()
```

```python
list(df_radon.columns)
```

```python
df_regbl.head()
```

```python
df = df_radon.merge(df_regbl, on="EGID", how="left")
```

```python
df.shape
```

```python
msno.matrix(df, labels=True, fontsize=8)
```

## Save the data

```python
df.to_csv(f"{ROOT}data/processed/dataset-v0.1.csv")
```
