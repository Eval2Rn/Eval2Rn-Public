# 0.8 - Process Minergie

This notebook's objective is to process the minergie dataset: `18_MINERGIE_buildings`

## Constants

```python
ROOT = '../../'
```

## Load the data

```python
gdf_minergie = gpd.read_file(f"{ROOT}data/raw/18_MINERGIE_buildings/minergie.gpkg")
```

```python
gdf_minergie.head()
```

```python
list(gdf_minergie.columns)
```

## Set correct CRS

```python
gdf_minergie.crs
```

```python
gdf_minergie.set_crs(epsg=2056, inplace=True, allow_override=True)
gdf_minergie.crs
```

```python
gdf_minergie.to_crs("EPSG:2056", inplace=True)
```

```python
gdf_minergie.head()
```

## Select only the needed columns

```python
list(gdf_minergie.columns)
```

```python
cols = [
 'certificate',
 'standard',
 'geometry'
]

gdf_minergie = gdf_minergie[cols]
```

## Transform column names with descriptive names

```python
gdf_minergie = gdf_minergie.rename(columns={
    'certificate': 'Miner - Certification de minergie',
    'standard': 'Miner - Standard de minergie'
})

list(gdf_minergie.columns)
```

## Transform geometry to coordinates

```python
gdf_minergie['Coordonnée E'] = gdf_minergie.geometry.x
gdf_minergie['Coordonnée N'] = gdf_minergie.geometry.y

# Drop geometry column
gdf_minergie = gdf_minergie.drop(columns=['geometry'])
```

```python
gdf_minergie.head()
```

```python
list(gdf_minergie.columns)
```

## Save the data

```python
gdf_minergie.to_csv(f"{ROOT}data/processed/02_MINERGIE_buildings.csv", index=False)
```
